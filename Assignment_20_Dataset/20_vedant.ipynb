```python
# Starting the code for the assignment / Importing main libraries
import pandas as pd  #csv Imported
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
df = pd.read_csv('ford_car_dataset.csv')
df #Prints the DataFrame
```

```python
# Question 1 (Preparing Features)
# Dependent Feature
y = df["price"]

# Independent Features
X = df.drop(columns=["price"])

print("The Shape of X is:", X.shape)
print("The Shape of y is:", y.shape)
```

```python
# Question 2 (One-Hot Encoding)
categorical_columns = X.select_dtypes(include=["object"]).columns

print("Categorical Columns:")
print(categorical_columns)

X_encoded = pd.get_dummies(X, columns=categorical_columns)

X_encoded = X_encoded.astype(int)

print("\nFirst 5 Rows After Encoding:")
print(X_encoded.head())
```

```python
# Question 3 (Feature Scaling)

scaler = StandardScaler()

numerical_columns = ["year", "mileage", "tax", "mpg", "engineSize"]

X_encoded[numerical_columns] = scaler.fit_transform(X_encoded[numerical_columns])

print("First 5 Rows of Scaled Features:")
print(X_encoded.head())
```

```python
# Question 4 (Train-Test Split)

X_train, X_test, y_train, y_test = train_test_split(X_encoded, y, test_size=0.33, random_state=42)

print("Shape of X_train:", X_train.shape)
print("Shape of X_test:", X_test.shape)
print("Shape of y_train:", y_train.shape)
print("Shape of y_test:", y_test.shape)
```

```python
# Question 5 (Building Linear Regression Model)

Lr_model = LinearRegression()

Lr_model.fit(X_train, y_train)

print("Intercept:", Lr_model.intercept_)

print("\nCoefficients from ford model:")
print(Lr_model.coef_)
```

```python
# Question 6 (Making Predictions)

y_predicted = Lr_model.predict(X_test)

print("First 10 Predicted Values:")
print(y_predicted[:10])

print("\nFirst 10 Actual Values:")
print(y_test.head(10))
```

```python
# Question 7 (Model Evaluation using R² Score)

r2 = r2_score(y_test, y_predicted)

print("R² Score:", r2)
```

### Model Performance

- The Linear Regression model is performing well.
- With an **R² score of 0.84**, the model explains about **84% of the variation** in car prices.
- Most of the predicted prices are close to the actual values.
- Overall, the model provides reliable and reasonably accurate price predictions.

```python
# Question 8 (Saving the Model)

import joblib

joblib.dump(Lr_model, "LR_ford_car.pkl")
joblib.dump(scaler, "scaler.pkl")
joblib.dump(X_encoded.columns.tolist(), "columns.pkl")

print("Model and preprocessing objects saved successfully.")
```
