```python
# Starting the code for the assignment / Importing main libraries
import pandas as pd  #csv Imported
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('apple_products_pricing_2020_2026.csv')

df #Prints
```

```python
#Question 1 (Head)
df.head(10) #First 10 rows of the dataset
```

```python
# Question 2 (shape, info, describe)

print("Dimensions :", df.shape)     # (Rows x Columns)

print("\n\nCol Name & dType :")
df.info()    #(Columns, DataType)

print("\n\nSummary :\n",df.describe())     #(Summary of the dataset)
```

```python
#Question 3 (Null Values)

print("Checking if there are any null values :\n\n",df.isnull().sum())   

#(Checking if there are any null values)
```

```python
# Question 4 (Numerical and Categorical Columns)

print("The Columns which Numerical data :")
print(df.select_dtypes(include=['float64', 'int64']).columns.tolist())

print("\n")

print("The columns which Categorical data :")
print(df.select_dtypes(include=['object']).columns.tolist())
```

```python
#Question 5 (Distribution Plots)               (color = '---color name---')
sns.histplot(df['Launch_Price_INR'],color='blue')
plt.title("Launch Price Inr Distribution")
plt.show()

sns.histplot(df['Rating'],color='green')
plt.title("Rating Distribution")
plt.show()

sns.histplot(df['Reviews_Count'],color='red')
plt.title("Number of Reviews Distribution")
plt.show()

sns.histplot(df['Discount_Pct'],color='orange')
plt.title("Discount Percentage Distribution")
plt.show()
```

```python
 # Question 6 (Countplot)
data=df
sns.countplot(x=data["Stock_Status"],hue=data["Condition"])
plt.title("Stock Status vs Condition ratio ")
plt.xlabel("Stock Status")
plt.ylabel("Count")
plt.show()

sns.countplot(x=data["Stock_Status"],hue=data["Platform"])
plt.title("Stock Status vs Platform Ratio")
plt.xlabel("Stock Status")
plt.ylabel("Count")
plt.show()

sns.countplot(x=data["Product_Category"],hue=data["Rating"])
plt.title("Product Category vs Rating Ratio")
plt.xlabel("Product Category")
plt.ylabel("Count")
plt.show()
```

```python
#Question 7 (Correlation Heatmap)
corr_Mat = df[["Launch_Price_INR", "Current_Price_INR", "Rating", "Discount_Pct"]].corr()
corr_Mat
sns.heatmap(corr_Mat, annot=True,cmap='Spectral')
plt.title("Correlation Heatmap \n")
```

```python
#Question 8 (Avg , Max , Min , Avg By Smoker)
avg_rating = df["Rating"].mean()
print("The average rating are :",avg_rating)

max_rating=df["Rating"].max()
print("\nThe Max rating are :",max_rating)

min_rating=df["Rating"].min()
print("\nThe min rating are :",min_rating)

avg_rating = df.groupby("Rating")["Reviews_Count"].mean()
print("\nAverage rating of a product with its review count:\n", avg_rating)


```

```python
# Question 9 (Boxplot)
plt.figure(figsize=(12, 10))

plt.subplot(2, 2, 1)
sns.boxplot(x=df["Product_Category"], y=df["Current_Price_INR"])
plt.title("Current_Price_INR Price by Category")

plt.subplot(2, 2, 2)
sns.boxplot(x=df["Rating"], y=df["Current_Price_INR"])
plt.title("Current_Price_INR Price by Rating")

plt.subplot(2, 2, 3)
sns.boxplot(x=df["Product_Category"], y=df["Current_Price_INR"], hue=df["Rating"])
plt.title("\nCurrent_Price_INR Price by Category and Rating")

plt.tight_layout()
plt.show()

```

```python
#Question 10 (Avg Launch Price and Avg Discount Percentage)
avg_Launch_Price = df["Launch_Price_USD"].mean()
avg_Discount_Pct = df["Discount_Pct"].mean()

print("Avg Launch Price :", avg_Launch_Price)
print("Avg Discount Percentage :", avg_Discount_Pct)

print("And So Yes The avg Launch Price of the product is", avg_Launch_Price,""
                    "and Average Discount Percentage of", avg_Discount_Pct, " ")
```

```python
# Question 10 (Average Launch Price, Average Discount, Average Reviews by Rating, Current Price Counts)

avg_Launch_Price = df["Launch_Price_USD"].mean()
avg_Discount_Pct = df["Discount_Pct"].mean()

avg_reviews = df.groupby("Rating")["Reviews_Count"].mean()

current_price_counts = df["Current_Price_USD"].value_counts()

print("Average Launch Price:", avg_Launch_Price)
print("Average Discount Percentage:", avg_Discount_Pct)

print("\nAverage Review Count by Rating:")
print(avg_reviews)

print("\nCurrent Price Counts:")
print(current_price_counts)
```

```python
# Question 10 (Product with Highest Rating)

max_rating_index = df["Rating"].idxmax()
product_data = df.loc[max_rating_index]

print("Full data of product with highest rating:\n")
print(product_data)
```
