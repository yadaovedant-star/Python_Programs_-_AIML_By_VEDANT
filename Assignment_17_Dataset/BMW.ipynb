## Dataset Summary

The dataset contains BMW car listings with the following attributes:

- **maker_key** – Manufacturer name
- **model_key** – Car model
- **mileage** – Distance traveled by the vehicle
- **engine_power** – Engine power rating
- **registration_date** – Date of vehicle registration
- **fuel** – Fuel type (e.g., Petrol, Diesel)
- **paint_color** – Exterior color of the vehicle
- **car_type** – Vehicle body type
- **feature_1** to **feature_8** – Binary indicators representing various vehicle features
- **price** – Selling price of the vehicle
- **sold_at** – Platform or source where the vehicle was sold

The dataset contains both **numerical** variables (such as *mileage*, *engine_power*, and *price*) and **categorical** variables (such as *fuel*, *car_type*, and *paint_color*).

This combination makes it suitable for **descriptive statistical analysis**, **data visualization**, and **machine learning models** aimed at analyzing and predicting BMW vehicle prices.

## Key Insights

1. **Fuel Type:** Most BMWs in the dataset are powered by **diesel**, followed by **petrol**, while **hybrid** and **electric** models are comparatively rare.

2. **Car Type:** **Estate**, **sedan**, and **SUV** are the most common BMW body styles, whereas **coupe**, **convertible**, and **van** models appear less frequently.

3. **Mileage vs Price:** Vehicles with **higher mileage** generally have **lower selling prices**, indicating a negative relationship between mileage and resale value.

4. **Engine Power vs Price:** Cars with **higher engine power** tend to have **higher selling prices**, suggesting that performance significantly influences market value.

5. **Paint Colors:** **Black**, **white**, and **grey** are the most common paint colors, with darker shades often showing a slightly higher resale value than other colors.

## Conclusion

The analysis of the BMW car dataset shows that **diesel** and **petrol** vehicles dominate the listings, with **estate**, **sedan**, and **SUV** being the most common body types. Vehicle **mileage** and **engine power** have a significant impact on the selling price, where higher mileage generally reduces price while greater engine power increases it. Additionally, **paint color** has a minor influence on resale value, with neutral shades such as **black**, **white**, and **grey** being the most popular. Overall, the dataset is well-suited for exploratory data analysis and predictive modeling of BMW vehicle prices.
