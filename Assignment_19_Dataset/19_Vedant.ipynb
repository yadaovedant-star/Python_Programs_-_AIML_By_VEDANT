```python
# Question 1 
# Starting the code for the assignment / Importing main libraries
import pandas as pd  #csv Imported
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
df = pd.read_csv('Updated_Mobile_Dataset.csv')
df = df.drop(columns=["Unnamed: 0"])

df #Prints the DataFrame
```

```python
# Question 2 

# Display first 10 rows
print(df.head(10))
```

- **Reason for Selection:** I selected the **Mobile Phones Dataset** because I am interested in smartphones and their specifications.

- **Dataset Quality:** The dataset is large, clean, and contains useful information such as brand, RAM, ROM, ratings, and price.

- **Target Variable:** I selected **Price** as the target variable because it is a **continuous numerical value**.

- **Purpose:** This dataset is suitable for performing Exploratory Data Analysis (EDA) and understanding the factors that influence mobile phone prices.

```python
# Question 3

print("Missing Values Count:")
print(df.isnull().sum())

percentage_is = (df.isnull().mean() * 100).round(2)

print("\nMissing Value Percentage:")
print(percentage_is)

df = df.dropna() # drops NAN values 

print("\nDuplicate Rows:", df.duplicated().sum()) #duplicates Count

df = df.drop_duplicates()

print("Duplicate Rows After Removal:", df.duplicated().sum())

print("\nFinal Dataset Shape:", df.shape)
```

### Observations

- Checked the dataset for missing values in each column.
- Calculated the percentage of missing values.
- Removed missing values using `dropna()`.
- Checked and removed duplicate rows using `drop_duplicates()`.
- The cleaned dataset was used for further analysis.

```python
# Question 4 

print(df.describe())

print("\nMinimum Price:", df["Price"].min())
print("Maximum Price:", df["Price"].max())
print("Mean Price:", df["Price"].mean())
print("Median Price:", df["Price"].median())
```

### Observations

- The minimum mobile phone price is much lower than the maximum price, showing a wide price range.
- The mean price is higher than the median price, indicating that a few expensive phones increase the average price.
- The dataset contains both budget and premium smartphones.
- The target variable **Price** is a continuous numerical variable, making it suitable for regression and statistical analysis.

```python
# Question 5 

plt.figure(figsize=(6,4))
sns.histplot(df["Price"], bins=20)
plt.title("Histogram of Price")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()

plt.figure(figsize=(6,4))
sns.histplot(df["Rating"], bins=10)
plt.title("Histogram of Rating")
plt.xlabel("Rating")
plt.ylabel("Frequency")
plt.show()
```

```python
#Question 9

print("Before Encoding")
print(df[["Company", "RamSize"]].head())

df_encoded = pd.get_dummies(df,columns=["Company", "RamSize", "RomSize"],dtype=int)

print("\nAfter Encoding")
print(df_encoded.head())
```

### Encoding Explanation

- One-Hot Encoding was used to convert categorical values into numeric form.
- **Company**, **RamSize**, and **RomSize** were encoded because they contain text data.
- Each category is represented using **0** and **1**, making it suitable for analysis and machine learning.
- The **Model** column was not encoded because it has many unique values, which would create too many unnecessary columns.
