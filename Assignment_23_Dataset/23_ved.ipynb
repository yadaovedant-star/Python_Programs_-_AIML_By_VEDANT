# Q1

```python
import pandas as pd

df_Lr = pd.read_csv("CarDetails.csv")

df_Lr
```

```python
df_Lr = df_Lr.drop_duplicates()
df_Lr = df_Lr.dropna()
```

```python
df_Lr.drop(columns=df_Lr.columns[0], inplace=True)

df_Encoded_Lr = pd.get_dummies(df_Lr)

df_Encoded_Lr = df_Encoded_Lr.astype(int)


df_Encoded_Lr
```

```python
from sklearn.model_selection import train_test_split

X = df_Encoded_Lr.drop("Price", axis=1)

y = df_Encoded_Lr["Price"]

X_train_Lr, X_test_Lr, y_train_Lr, y_test_Lr = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

```

```python
from sklearn.linear_model import LinearRegression

modelLr=LinearRegression()
modelLr.fit(X_train_Lr,y_train_Lr)

```

```python
y_pred_Lr = modelLr.predict(X_test_Lr)
y_pred_Lr
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y_test_Lr, y_pred_Lr)

r2
```

# Q3

```python
import pandas as pd

df_KNN = pd.read_csv("heart.csv")

df_KNN
```

```python
df_KNN = df_KNN.drop_duplicates()
df_KNN = df_KNN.dropna()
```

```python
df_KNN.drop(columns=df_KNN.columns[0], inplace=True)

df_Encoded_Knn= pd.get_dummies(df_KNN)

df_Encoded_Knn = df_Encoded_Knn.astype(int)


df_Encoded_Knn
```

```python
from sklearn.model_selection import train_test_split

X = df_Encoded_Knn.drop("HeartDisease", axis=1)

y = df_Encoded_Knn["HeartDisease"]

X_train_Knn, X_test_Knn, y_train_Knn, y_test_Knn = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train_Knn)
X_test_scaled = scaler.transform(X_test_Knn)
```

```python
from sklearn.neighbors import KNeighborsClassifier


modelKNN = KNeighborsClassifier(n_neighbors=3)

modelKNN.fit(X_train_scaled, y_train_Knn)
```

```python
from sklearn.metrics import accuracy_score

y_pred_Knn = modelKNN.predict(X_test_scaled)

print("Predictions:")
print(y_pred_Knn)

accuracy = accuracy_score(y_test_Knn, y_pred_Knn)

print("\nAccuracy:", accuracy)
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

for k in [1, 3, 5, 7, 9, 11, 13, 15]:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train_scaled, y_train_Knn)

    y_pred_Knn = model.predict(X_test_scaled)

    accuracy = accuracy_score(y_test_Knn, y_pred_Knn)

    print(f"k = {k}  =  Accuracy = {accuracy:}")
```

## Prediction and Model Evaluation

- The trained **K-Nearest Neighbors (KNN)** model is used to predict the class labels for the test dataset. 
- The predicted values indicate whether a patient is classified as **0 (No Heart Disease)** or **1 (Heart Disease)**. 
- After obtaining the predictions, the **Accuracy Score** is calculated to evaluate the model's performance. 
- Accuracy represents the proportion of correctly classified instances out of the total test samples 
- Also helps determine how well the model performs on unseen data.

# Q4

```python
import pandas as pd

df_Nb = pd.read_csv("heart.csv")

df_Nb
```

```python
df_Nb= df_Nb.drop_duplicates()
df_Nb = df_Nb.dropna()
```

```python
df_Nb.drop(columns=df_Nb.columns[0], inplace=True)

df_Encoded_Nb= pd.get_dummies(df_Nb)

df_Encoded_Nb = df_Encoded_Nb.astype(int)


df_Encoded_Nb
```

```python
from sklearn.naive_bayes import GaussianNB

from sklearn.model_selection import train_test_split


X = df_Encoded_Nb.drop("HeartDisease", axis=1)

y = df_Encoded_Nb["HeartDisease"]

X_train_NB, X_test_NB, y_train_NB, y_test_NB = train_test_split(X, y, test_size=0.20, random_state=42)
```

```python
modelNB = GaussianNB()

modelNB.fit(X_train_NB, y_train_NB)
```

```python
y_pred_NB = modelNB.predict(X_test_NB)

print("Predictions:")
print(y_pred_NB)
```

```python
from sklearn.metrics import classification_report

print("Classification Report")
print(classification_report(y_test_NB, y_pred_NB))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test_NB, y_pred_NB)

print("Confusion Matrix")
print(cm)
```

## Result

- The Gaussian Naive Bayes classifier was successfully trained and tested on the Heart Disease dataset. 
- The Classification Report provides the Precision, Recall, F1-Score, and Support for each class
- while the Confusion Matrix shows the number of correctly and incorrectly classified instances. 
- These metrics are used to evaluate the overall performance of the Naive Bayes classifier.
