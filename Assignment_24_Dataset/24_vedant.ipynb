Q1 ]
- Classification Dataset (Heart Disease)

```python
import pandas as pd

# Load Dataset
df_class = pd.read_csv("heart.csv")

# Display Dataset
print(df_class.head())

# Dataset Information
print(df_class.info())

# Check Missing Values
print(df_class.isnull().sum())

# Remove Missing Values
df_class = df_class.dropna()

# Remove Duplicate Rows
df_class = df_class.drop_duplicates()

# Reset Index
df_class.reset_index(drop=True, inplace=True)

# Replace invalid values with mean
columns_to_fix = ["RestingBP", "Cholesterol"]

for col in columns_to_fix:
    mean_value = df_class.loc[df_class[col] != 0, col].mean()
    df_class[col] = df_class[col].replace(0, mean_value)

```

```python
print("Duplicate Rows :", df_class.duplicated().sum())

print("\nRestingBP Zero Values :", (df_class["RestingBP"] == 0).sum())
print("Cholesterol Zero Values :", (df_class["Cholesterol"] == 0).sum())
```

```python

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler


# One-Hot Encoding
df_class_encoded = pd.get_dummies(df_class)

# Independent Features (X)
X_class = df_class_encoded.drop("HeartDisease", axis=1)

# Dependent Feature (y)
y_class = df_class_encoded["HeartDisease"]

```

```python
# Train-Test Split
X_train_class, X_test_class, y_train_class, y_test_class = train_test_split(
    X_class,
    y_class,
    test_size=0.2,
    random_state=42
)

# Feature Scaling
scaler_class = StandardScaler()

X_train_class = scaler_class.fit_transform(X_train_class)
X_test_class = scaler_class.transform(X_test_class)

print("X Shape :", X_class.shape)
print("y Shape :", y_class.shape)
print("X_train :", X_train_class.shape)
print("X_test  :", X_test_class.shape)
```

Q1 ]
- Regression Dataset (Car Price)

```python
import pandas as pd

# Load Dataset
df_reg = pd.read_csv("Car_Price_Prediction.csv")

# Display Dataset
print(df_reg.head())

# Dataset Information
df_reg.info()

# Check Missing Values
print(df_reg.isnull().sum())

# Check Duplicate Rows
print("Duplicate Rows :", df_reg.duplicated().sum())

# Remove Missing Values
df_reg.dropna(inplace=True)

# Remove Duplicate Rows
df_reg.drop_duplicates(inplace=True)

# Reset Index
df_reg.reset_index(drop=True, inplace=True)

# Display Cleaned Dataset
df_reg
```

```python
# One-Hot Encoding
df_reg_encoded = pd.get_dummies(df_reg)

# Independent Features (X)
X_reg = df_reg_encoded.drop("Price", axis=1)

# Dependent Feature (y)
y_reg = df_reg_encoded["Price"]
```

```python
from sklearn.model_selection import train_test_split


# Train-Test Split

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

print("X Shape :", X_reg.shape)
print("y Shape :", y_reg.shape)
print("X_train :", X_train_reg.shape)
print("X_test  :", X_test_reg.shape)



```

Q 2.1 ]
- (Classification Algorithms)

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

# Train Logistic Regression Model
modelLogi = LogisticRegression(max_iter=1000)

modelLogi.fit(X_train_class, y_train_class)

# Prediction
y_pred_Logi = modelLogi.predict(X_test_class)

# Evaluation
accuracy_Logi = accuracy_score(y_test_class, y_pred_Logi)

print("Accuracy :", accuracy_Logi)

print("\nConfusion Matrix")
print(confusion_matrix(y_test_class, y_pred_Logi))

print("\nClassification Report")
print(classification_report(y_test_class, y_pred_Logi))
```

Q 2.2 ]
-  (Regression Algorithms)

```python
from sklearn.tree import DecisionTreeClassifier

# Train Decision Tree Model
modelDT = DecisionTreeClassifier(random_state=42)

modelDT.fit(X_train_class, y_train_class)

# Prediction
y_pred_DT = modelDT.predict(X_test_class)

# Evaluation
accuracy_DT = accuracy_score(y_test_class, y_pred_DT)

print("Accuracy :", accuracy_DT)

print("\nConfusion Matrix")
print(confusion_matrix(y_test_class, y_pred_DT))

print("\nClassification Report")
print(classification_report(y_test_class, y_pred_DT))
```

Q 2.3 ]
- Support Vector Machine (SVM) 

```python
from sklearn.svm import SVC

# Train SVM Model
modelSVM = SVC()

modelSVM.fit(X_train_class, y_train_class)

# Prediction
y_pred_SVM = modelSVM.predict(X_test_class)

# Evaluation
accuracy_SVM = accuracy_score(y_test_class, y_pred_SVM)

print("Accuracy :", accuracy_SVM)

print("\nConfusion Matrix")
print(confusion_matrix(y_test_class, y_pred_SVM))

print("\nClassification Report")
print(classification_report(y_test_class, y_pred_SVM))
```

Q 2.4 ]
- K-Nearest Neighbors

```python
from sklearn.neighbors import KNeighborsClassifier

# Train KNN Model
modelKNN = KNeighborsClassifier(n_neighbors=5)

modelKNN.fit(X_train_class, y_train_class)

# Prediction
y_pred_KNN = modelKNN.predict(X_test_class)

# Evaluation
accuracy_KNN = accuracy_score(y_test_class, y_pred_KNN)

print("Accuracy :", accuracy_KNN)

print("\nConfusion Matrix")
print(confusion_matrix(y_test_class, y_pred_KNN))

print("\nClassification Report")
print(classification_report(y_test_class, y_pred_KNN))
```
