```python
# Starting the code for the assignment / Importing main libraries
import pandas as pd  #csv Imported
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('insurance.csv')

df #Prints
```

```python
#Question 1 (Head)
df.head(10) #First 10 rows of the dataset
```

```python
# Question 2 (shape, info, describe)

print("Dimensions :", df.shape)     # (Rows x Columns)

print("\n\nCol Name & dType :")
df.info()    #(Columns, DataType)

print("\n\nSummary :\n",df.describe())     #(Summary of the dataset)
```

```python
#Question 3 (Null Values)

print("Checking if there are any null values :\n\n",df.isnull().sum())   

#(Checking if there are any null values)
```

```python
# Question 4 (Numerical and Categorical Columns)

print("The Columns which Numerical data :")
print(df.select_dtypes(include=['float64', 'int64']).columns.tolist())

print("\n")

print("The columns which Categorical data :")
print(df.select_dtypes(include=['object']).columns.tolist())
```

```python
#Question 5 (Distribution Plots)               (color = '---color name---')
sns.histplot(df['age'],color='blue')
plt.title("Age Distribution")
plt.show()

sns.histplot(df['bmi'],color='green')
plt.title("BMI Distribution")
plt.show()

sns.histplot(df['children'],color='red')
plt.title("Number of Children Distribution")
plt.show()

sns.histplot(df['expenses'],color='orange')
plt.title("Medical Expenses Distribution")
plt.show()
```

```python
 # Question 6 (Countplot)
data=df
sns.countplot(x=data["sex"],hue=data["smoker"])
plt.title("Sex Ratio vs Smoker ratio ")
plt.xlabel("sex")
plt.ylabel("Count")
plt.show()

sns.countplot(x=data["smoker"],hue=data["region"])
plt.title("Smokers Ratio vs Region Ratio")
plt.xlabel("Smoker")
plt.ylabel("Count")
plt.show()

sns.countplot(x=data["region"],hue=data["sex"])
plt.title("Region Ratio vs Sex ratio")
plt.xlabel("region")
plt.ylabel("Count")
plt.show()
```

```python
#Question 7 (Correlation Heatmap)
corr_Mat = df[["age", "bmi", "children", "expenses"]].corr()
corr_Mat
sns.heatmap(corr_Mat, annot=True,cmap='Spectral')
plt.title("Correlation Heatmap For Numerical Data / column \n")
```

```python
#Question 8 (Avg , Max , Min , Avg By Smoker)
avg_charges = df["expenses"].mean()
print("The average charges are :",avg_charges)

max_charges=df["expenses"].max()
print("\nThe Max charges are :",max_charges)

min_charges=df["expenses"].min()
print("\nThe min charges are :",min_charges)

avg_in_case_of_smoker = df.groupby("smoker")["expenses"].mean()
print("\nAverage expense of a smokers and non-smokers:", avg_in_case_of_smoker)
```

```python
# Question 9 (Boxplot)

plt.figure(figsize=(12, 10))


plt.subplot(2, 2, 1)
sns.boxplot(x=df["smoker"], y=df["expenses"])
plt.title("Charges by Smoker")


plt.subplot(2, 2, 2)
sns.boxplot(x=df["sex"], y=df["expenses"])
plt.title("Charges by Sex")

plt.subplot(2, 2, 3)
sns.boxplot(x=df["smoker"], y=df["expenses"], hue=df["sex"])
plt.title("\nCharges by Smoker and Sex Using Hue Parameter")

plt.tight_layout()
plt.show()

```

```python
#Question 10 (Avg Age and Avg BMI)
avg_age = df["age"].mean()
avg_bmi = df["bmi"].mean()

print("Avg Age :", avg_age)
print("Avg Body Mass Index (BMI) :", avg_bmi)

print("And So Yes The avg Age of the customer is", avg_age.round(0),""
                    "and Average BMI of", avg_bmi.round(0), " ")
```

```python
##Question 10 (Avg Age , Avg BMI , Smoker vs Non-Smoker Charges , Region Counts)
avg_age = df["age"].mean().round(0)
avg_bmi = df["bmi"].mean().round(0)

smoker_charges = df.groupby("smoker")["expenses"].mean()
region_counts = df["region"].value_counts()

print("Average Age:", avg_age)
print("Average BMI:", avg_bmi)

print("\nSmoker vs Non-Smoker Charges:")
print("Non-Smoker charges:", smoker_charges["no"])
print("Smoker charges:", smoker_charges["yes"])

print("\nRegion Counts:")
print(region_counts)
```

```python
#Question 10 (Person with Highest Insurance Charges)
# Find person with highest insurance charges
max_expense_index = df["expenses"].idxmax()
person_data = df.loc[max_expense_index]

print("Full data of person with highest insurance charges:\n")
print(person_data)

```

```python
# Question 5 (Distribution Plots)

# Rank Distribution
sns.histplot(df['rank'], color='blue')
plt.title("Rank Distribution")
plt.show()

# Rating Distribution
sns.histplot(df['rating'], color='green')
plt.title("Rating Distribution")
plt.show()

# Vote Count Distribution
sns.histplot(df['vote_count'], color='red')
plt.title("Vote Count Distribution")
plt.show()

# Variance Distribution
sns.histplot(df['variance'], color='orange')
plt.title("Variance Distribution")
plt.show()
```

```python
# Question 6 (Countplot with Limited Data)

# Limit dataset to first 20 rows
data = df.head(20)

# Rank vs Category
sns.countplot(x="rank", hue="category", data=data)
plt.title("Rank vs Category Ratio (First 20 Items)")
plt.xlabel("Rank")
plt.ylabel("Count")
plt.show()

# License vs Organization
sns.countplot(x="license", hue="organization", data=data)
plt.title("License vs Organization Ratio (First 20 Items)")
plt.xlabel("License")
plt.ylabel("Count")
plt.xticks(rotation=45)
plt.show()

# Model Name vs Rating (First 20 Items)
sns.countplot(x="model_name", hue="rating", data=data)
plt.title("Model Name vs Rating Ratio (First 20 Items)")
plt.xlabel("Model Name")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()

```

```python
# Question 7 (Correlation Heatmap)

# Select only numeric columns for correlation
corr_Mat = df[["rating", "variance", "vote_count", "rank"]].corr()

# Display correlation matrix
print(corr_Mat)

# Plot heatmap
sns.heatmap(corr_Mat, annot=True, cmap='Spectral')
plt.title("Correlation Heatmap (Numeric Columns)")
plt.show()
```

```python
# Question 8 (Avg, Max, Min, Avg By Rating)

# Average rating
avg_rating = df["rating"].mean()
print("The average rating is :", avg_rating)

# Maximum rating
max_rating = df["rating"].max()
print("\nThe maximum rating is :", max_rating)

# Minimum rating
min_rating = df["rating"].min()
print("\nThe minimum rating is :", min_rating)

# Average vote count grouped by rating
avg_votes_by_rating = df.groupby("rating")["vote_count"].mean()
print("\nAverage vote count for each rating:\n", avg_votes_by_rating)


```

```python
# Question 10 (Average values from numeric columns)

# Average Rating
avg_rating = df["rating"].mean()

# Average Lower Rating
avg_rating_lower = df["rating_lower"].mean()

# Average Upper Rating
avg_rating_upper = df["rating_upper"].mean()

# Average Variance
avg_variance = df["variance"].mean()

# Average Vote Count
avg_vote_count = df["vote_count"].mean()

# Average Rank
avg_rank = df["rank"].mean()

# Print results
print("Average Rating:", avg_rating)
print("Average Lower Rating:", avg_rating_lower)
print("Average Upper Rating:", avg_rating_upper)
print("Average Variance:", avg_variance)
print("Average Vote Count:", avg_vote_count)
print("Average Rank:", avg_rank)

# Summary statement
print(f"\nSo yes, the dataset shows an average rating of {avg_rating}, "
      f"\naverage variance of {avg_variance}, \naverage vote count of {avg_vote_count}, "
      f"\nand average rank of {avg_rank}.")
```

```python
# Question 10 (Average values and counts with valid columns)

# Average Rating
avg_rating = df["rating"].mean()

# Average Variance
avg_variance = df["variance"].mean()

# Average Vote Count grouped by Rating
avg_votes_by_rating = df.groupby("rating")["vote_count"].mean()

# Rank counts (frequency of each rank value)
rank_counts = df["rank"].value_counts()

# Print results
print("Average Rating:", avg_rating)
print("Average Variance:", avg_variance)

print("\nAverage Vote Count by Rating:")
print(avg_votes_by_rating)

print("\nRank Counts:")
print(rank_counts)
```
